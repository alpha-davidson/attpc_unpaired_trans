```python
import h5py
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import random
```

```python
filename = 'fission_data/Bi200.h5'
with h5py.File(filename, 'r') as f:
    keys = [k for k in f if k.startswith("Event_")]
    selected = random.sample(keys, 6)

    fig = plt.figure(figsize=(18, 10))
    for i, key in enumerate(selected):
        hit = f[key]["HitArray"][()]
        x, y, z, charge = hit['x'], hit['y'], hit['z'], hit['A']

        ax = fig.add_subplot(2, 3, i+1, projection='3d')
        sc = ax.scatter(z, x, y, c=charge, cmap='viridis', s=5)

        ax.set_title(key)
        ax.set_xlabel("z (mm)")
        ax.set_ylabel("x (mm)")    
        ax.set_zlabel("y (mm)")       
        ax.set_xlim(0, 1000)     
        ax.set_ylim(-250, 250)      
        ax.set_zlim(-250, 250)       

        ax.set_box_aspect([1, 1, 1])
        ax.grid(False)
        ax.view_init(elev=20, azim=-60)
    cbar_ax = fig.add_axes([0.92, 0.15, 0.015, 0.7])
    fig.colorbar(sc, cax=cbar_ax, label='Charge (A)')

    plt.tight_layout(rect=[0, 0, 0.9, 1])
    plt.show()
```

```python
filename = 'fission_data/Bi200SimMoreTriggered.h5'
event_indices = [1308, 1512, 2976, 2964, 4633, 4349]
event_keys = [f"Event_[{idx}]" for idx in event_indices]

with h5py.File(filename, "r") as f:
    fig = plt.figure(figsize=(18, 10))

    for i, key in enumerate(event_keys):
        if key not in f:
            print(f"Warning: {key} not found in file.")
            continue

        hit_array = f[key]["HitArray"][()]
        x = hit_array['x']
        y = hit_array['y']
        z = hit_array['z']
        charge = hit_array['A']

        ax = fig.add_subplot(2, 3, i + 1, projection='3d')
        sc = ax.scatter(z, x, y, c=charge, cmap='viridis', s=5)
        ax.set_title(f"{key}")
        ax.set_xlabel("z (mm)")
        ax.set_ylabel("x (mm)")
        ax.set_zlabel("y (mm)")

        ax.set_xlim(0, 1000) 
        ax.set_ylim(-250, 250) 
        ax.set_zlim(-250, 250)     

        ax.set_box_aspect([1, 1, 1])
        ax.grid(False)
        ax.view_init(elev=20, azim=-60)

    cbar_ax = fig.add_axes([0.92, 0.15, 0.015, 0.7])
    fig.colorbar(sc, cax=cbar_ax, label='Charge (A)')

    plt.tight_layout(rect=[0, 0, 0.9, 1])
    plt.show()
```
