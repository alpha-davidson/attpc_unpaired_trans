```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

path = '../fission_data/Fission_sim.npy'
path2 = '../fission_data/Fission_sc.npy'
data = np.load(path)
data2 = np.load(path2)
```

```python
num_samples = 6
fig = plt.figure(figsize=(18, 10))

for i in range(num_samples):
    sample = data[i]
    x, z, y, amp = sample[:, 0], sample[:, 1], sample[:, 2], sample[:, 3]
    
    ax = fig.add_subplot(2, 3, i + 1, projection='3d')
    sc = ax.scatter(x, y, z, c=amp, cmap='plasma', s=3)
    ax.set_title(f"Sample {i}")
    ax.set_xlabel("x")
    ax.set_ylabel("z")
    ax.set_zlabel("y")
    ax.set_zlim(-1, 1)
    ax.set_ylim(-1, 1)
    ax.set_xlim(-1, 1)

fig.colorbar(sc, ax=fig.get_axes(), shrink=0.5, label='Amplitude')
plt.show()
```

```python
num_samples = 6
fig = plt.figure(figsize=(18, 10))

for i in range(num_samples):
    sample = data2[i]
    x, z, y, amp = sample[:, 0], sample[:, 1], sample[:, 2], sample[:, 3]
    
    ax = fig.add_subplot(2, 3, i + 1, projection='3d')
    sc = ax.scatter(x, y, z, c=amp, cmap='plasma', s=3)
    ax.set_title(f"Sample {i}")
    ax.set_xlabel("x")
    ax.set_ylabel("z")
    ax.set_zlabel("y")
    ax.set_zlim(-1, 1)
    ax.set_ylim(-1, 1)
    ax.set_xlim(-1, 1)

fig.colorbar(sc, ax=fig.get_axes(), shrink=0.5, label='Amplitude')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Compute z differences across all samples
z_diffs = []

for sample1, sample2 in zip(data, data2):
    z1 = sample1[:, 1]  # assuming z is the second column
    z2 = sample2[:, 1]
    
    # Subtract corresponding z-values
    diff = z1 - z2
    z_diffs.append(diff)

# Flatten into one array
z_diffs = np.concatenate(z_diffs)

# Plot histogram
plt.figure(figsize=(8, 6))
plt.hist(z_diffs, bins=100, color='teal', alpha=0.8)
plt.title("Histogram of z-value differences (data - data2)")
plt.xlabel("Δz")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```
